# Task 39. Exercise on Numpy Array sorting

```python
# Sorting Arrays
# Sort the array:
import numpy as np

arr = np.array([3, 2, 0, 1])

print(np.sort(arr))
```

```python
# Sort the array alphabetically:
import numpy as np

arr = np.array(['banana', 'cherry', 'apple'])

print(np.sort(arr))
```

```python
# Sort a boolean array:
import numpy as np

arr = np.array([True, False, True])

print(np.sort(arr))
```

```python
# Sorting a 2-D Array
# Sort a 2-D array:
import numpy as np

arr = np.array([[3, 2, 4], [5, 0, 1]])

print(np.sort(arr))
```
